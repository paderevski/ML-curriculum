## 2: Operators: Arithmetic and String Concatenation

Basic math operators, and `+` for joining strings together.

```python
# Arithmetic operations
sum_result = 10 + 5
print("Sum:", sum_result)

difference = 10 - 5
print("Difference:", difference)

product = 10 * 5
print("Product:", product)

division = 10 / 5
print("Division:", division)
```

## 3: Lists and Dictionaries

Two of the most common data structures: ordered lists and key-value
dictionaries.

```python
# Lists (ordered, mutable)
fruits = ["apple", "banana", "cherry"]
print("Fruits:", fruits)
print("First fruit:", fruits[0])
```

```python
# Dictionaries (unordered, mutable, key-value pairs)
person = {"name": "Bob", "age": 30}
print("Person:", person)
print("Bob's age:", person["age"])
```

## 5: Loops

## 8: Matplotlib Samples

A quick tour of the plot types you'll use constantly this year: line plots,
scatter plots, bar charts, histograms, subplots, and a first look at 3D.

### 8.1 Line Plot

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 6))
plt.plot(x, y, label='sin(x)')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Basic Line Plot')
plt.legend()
plt.grid(True)
plt.show()
```

### 8.2 Scatter Plot

```python
np.random.seed(0)
x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)
colors = np.random.rand(50)
sizes = 1000 * np.random.rand(50)

plt.figure(figsize=(8, 6))
plt.scatter(x_scatter, y_scatter, c=colors, s=sizes, alpha=0.5, cmap='viridis')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.colorbar(label='Color intensity')
plt.show()
```

### 8.3 Bar Chart

```python
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 12]

plt.figure(figsize=(8, 6))
plt.bar(categories, values, color='skyblue')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart')
plt.show()
```

### 8.4 Histogram

```python
data = np.random.randn(1000)

plt.figure(figsize=(8, 6))
plt.hist(data, bins=30, color='lightgreen', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

### 8.5 Subplots

```python
x_sub = np.linspace(0, 10, 100)
y1_sub = np.sin(x_sub)
y2_sub = np.cos(x_sub)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(x_sub, y1_sub, color='red')
axes[0].set_title('Sine Wave')
axes[0].set_xlabel('x')
axes[0].set_ylabel('sin(x)')

axes[1].plot(x_sub, y2_sub, color='purple')
axes[1].set_title('Cosine Wave')
axes[1].set_xlabel('x')
axes[1].set_ylabel('cos(x)')

plt.tight_layout()
plt.show()
```

### 8.6 Customizing Styles

```python
plt.style.use('seaborn-v0_8-darkgrid')  # Example of using a style

x_style = np.linspace(0, 10, 100)
y_style = np.sin(x_style)

plt.figure(figsize=(8, 6))
plt.plot(x_style, y_style, linestyle='--', color='orange', linewidth=2)
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Styled Line Plot')
plt.show()

# You can list available styles with:
# print(plt.style.available)
```

### 8.7 3D Plot

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

x_3d = np.linspace(-5, 5, 50)
y_3d = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x_3d, y_3d)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

ax.plot_surface(X, Y, Z, cmap='viridis')

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('3D Surface Plot')

plt.show()
```
